```scala
import org.apache.spark.sql.SparkSession

val spark = SparkSession.builder()
  .remote("sc://spark-connect.kubeflow-user-example-com.svc.cluster.local:15002")
  .getOrCreate()

println(s"Spark version: ${spark.version}")
```

```scala
import spark.implicits._

spark.range(0, 1000)
  .withColumn("bucket", ($"id" % 10).cast("string"))
  .groupBy("bucket")
  .count()
  .orderBy("bucket")
  .show()
```

```scala
import org.apache.spark.sql.functions.{rand, sum}

val df = spark.range(0, 20000000)
  .repartition(2000)
  .withColumn("r", rand())
  .withColumn("bucket", ($"id" % 2000).cast("string"))

println(df.groupBy("bucket").agg(sum("r")).count())
```
